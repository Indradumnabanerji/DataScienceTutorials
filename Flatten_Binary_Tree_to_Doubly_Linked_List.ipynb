```
# This is the class for a Binary Tree node.
# It's assumed to be provided by the environment.
class BinaryTree:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

# This function flattens a Binary Tree into a Doubly Linked List in-place.
# The 'left' pointer becomes 'prev' and 'right' becomes 'next'.
# It returns the leftmost node of the flattened list.
def flattenBinaryTree(root):
    """
    Flattens a Binary Tree into a Doubly Linked List in-place,
    following the original tree's left-to-right (in-order) traversal order.
    The 'left' pointer acts as 'prev' and 'right' as 'next'.

    Args:
        root (BinaryTree): The root node of the Binary Tree to flatten.

    Returns:
        BinaryTree: The leftmost node (head) of the flattened Doubly Linked List.
    """
    # We use a recursive helper function that returns the head and tail of the
    # flattened linked list for the current subtree.

    # The main function simply calls the helper and returns the head.
    head, _ = flatten_tree_helper(root)
    return head

def flatten_tree_helper(node):
    """
    Recursive helper function to flatten a subtree rooted at 'node'.
    Returns a tuple: (head_of_flattened_list, tail_of_flattened_list)
    """
    # Base case: If the node is None, there's no list to flatten.
    if node is None:
        return None, None

    # Recursively flatten the left subtree.
    # left_head: head of the flattened left list
    # left_tail: tail of the flattened left list
    left_head, left_tail = flatten_tree_helper(node.left)

    # Recursively flatten the right subtree.
    # right_head: head of the flattened right list
    # right_tail: tail of the flattened right list
    right_head, right_tail = flatten_tree_helper(node.right)

    # --- Process the current node ---
    # The current node will be the new head if there's no left subtree.
    # Otherwise, the head of the left subtree is the overall head for this segment.
    head = left_head if left_head else node
    tail = right_tail if right_tail else node

    # Connect the left subtree's tail to the current node.
    # If there's a left subtree, its tail's 'right' (next) pointer should point to the current node.
    # The current node's 'left' (prev) pointer should point to the left subtree's tail.
    if left_tail:
        left_tail.right = node
        node.left = left_tail
    else:
        # If no left subtree, the current node's left pointer should be None.
        node.left = None

    # Connect the current node to the right subtree's head.
    # The current node's 'right' (next) pointer should point to the right subtree's head.
    # If there's a right subtree, its head's 'left' (prev) pointer should point to the current node.
    if right_head:
        node.right = right_head
        right_head.left = node
    else:
        # If no right subtree, the current node's right pointer should be None.
        node.right = None

    # Return the head and tail of the completely flattened list for this subtree.
    return head, tail
```
